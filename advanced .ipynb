```python
import sqlite3
conn=sqlite3.connect("col.db")
cursor=conn.cursor()
cursor.execute("""DROP TABLE IF EXISTS marks""")
cursor.execute("""DROP TABLE IF EXISTS stud""")
cursor.execute("""CREATE TABLE IF NOT EXISTS stud(stud_id INTEGER PRIMARY KEY,stu_name TEXT NOT NULL,dept TEXT,admission_year INTEGER)""")
conn.commit()
cursor.execute("""CREATE TABLE IF NOT EXISTS marks(mark_id INTEGER PRIMARY KEY,stud_id INTEGER,subject TEXT,mark INTEGER, FOREIGN KEY (stud_id) REFERENCES stud(stud_id))""")
conn.commit()
stud_data=[(1,"ram","EEE",2021),(2,"sam","ECE",2021),(3,"ravi","IT",2023)]
marks_data=[(11,1,"Mechanical",85),(21,2,"Physics",76),(22,3,"science",56)]
cursor.executemany("""INSERT INTO stud VALUES(?,?,?,?)""",stud_data)
cursor.executemany("""INSERT INTO marks VALUES(?,?,?,?)""",marks_data)
conn.commit()
cursor.execute("""UPDATE marks SET mark = mark + 5 WHERE stud_id IN (SELECT stud_id FROM stud WHERE admission_year < 2022)""")
conn.commit()
cursor.execute("""DELETE FROM marks WHERE mark < 35 AND stud_id IN (SELECT stud_id FROM marks GROUP BY stud_id HAVING COUNT(subject) = 2)""")
conn.commit()
```

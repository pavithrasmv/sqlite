```python
#ACADEMIC RECORDS INTEGRITY AND ANALYSIS
import sqlite3
conn=sqlite3.connect("col.db")
cursor=conn.cursor()
cursor.execute("""DROP TABLE IF EXISTS marks""")
cursor.execute("""DROP TABLE IF EXISTS stud""")
cursor.execute("""CREATE TABLE IF NOT EXISTS stud(stud_id INTEGER PRIMARY KEY,stu_name TEXT NOT NULL,dept TEXT,admission_year INTEGER)""")
conn.commit()
cursor.execute("""CREATE TABLE IF NOT EXISTS marks(mark_id INTEGER PRIMARY KEY,stud_id INTEGER,subject TEXT,mark INTEGER, FOREIGN KEY (stud_id) REFERENCES stud(stud_id))""")
conn.commit()
stud_data=[(1,"ram","EEE",2021),(2,"sam","ECE",2021),(3,"ravi","IT",2023)]
marks_data=[(11,1,"Mechanical",85),(21,2,"Physics",76),(22,3,"science",56)]
cursor.executemany("""INSERT INTO stud VALUES(?,?,?,?)""",stud_data)
cursor.executemany("""INSERT INTO marks VALUES(?,?,?,?)""",marks_data)
conn.commit()
cursor.execute("""UPDATE marks SET mark = mark + 5 WHERE stud_id IN (SELECT stud_id FROM stud WHERE admission_year < 2022)""")
conn.commit()
cursor.execute("""DELETE FROM marks WHERE mark < 35 AND stud_id IN (SELECT stud_id FROM marks GROUP BY stud_id HAVING COUNT(subject) = 2)""")
conn.commit()
cursor.execute("""SELECT dept,AVG(mark),count(*)AS tot_stu from stud INNER JOIN marks ON stud.stud_id=marks.stud_id GROUP BY dept""")
conn.commit()
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

```python
#EMPLOYEE MANAGEMENT AND SALARY CONTROL
import sqlite3
conn=sqlite3.connect("Empl.db")
cursor=conn.cursor()
cursor.execute("""DROP TABLE IF EXISTS dept""")
cursor.execute("""DROP TABLE IF EXISTS emplo""")
cursor.execute("""CREATE TABLE IF NOT EXISTS dept(dept_id PRIMARY KEY,dept_name UNIQUE NOT NULL)""")
cursor.execute("""CREATE TABLE IF NOT EXISTS emplo(emp_id PRIMARY KEY,emp_name TEXT,dept_id INTEGER,salary INTEGER,joining_date DATE,FOREIGN KEY (dept_id) REFERENCES dept(dept_id))""")
dept_data=[(1,"ECE"),(2,"EEE"),(3,"IT"),(4,"CSE")]
emp_data=[(12,"ram",1,50000,'2015'),(13,"sam",2,40000,'2014'),(14,"sanu",3,35000,'2016'),(15,"tena",4,45000,'2017')]
cursor.executemany("""INSERT INTO dept VALUES(?,?)""",dept_data)
cursor.executemany("""INSERT INTO emplo VALUES(?,?,?,?,?)""",emp_data)
conn.commit()
cursor.execute("SELECT * FROM dept")
rows=cursor.fetchall()
for row in rows:
  print(row)
cursor.execute("SELECT * FROM emplo")
rows=cursor.fetchall()
for row in rows:
  print(row)
cursor.execute("""UPDATE emplo SET salary=(salary*0.10) WHERE joining_date='2015' AND joining_date > 5""")
conn.commit()
cursor.execute("""UPDATE emplo SET salary=(salary*0.95) WHERE dept_id in(SELECT dept_id FROM emplo group by dept_id HAVING AVG(salary)>80000 )""")
cursor.execute("""SELECT dept_name ,count(salary)AS tot_salary,MAX(salary), dept.dept_id FROM dept INNER JOIN emplo ON dept.dept_id =emplo.dept_id GROUP BY dept_name""")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

```python
#SALES SYSTEM WITH REVENUE AND RISK ANALYSIS
import sqlite3
conn=sqlite3.connect("sale3.db")
cursor=conn.cursor()
cursor.execute("""DROP TABLE IF EXISTS cust""")
cursor.execute("""DROP TABLE IF EXISTS ord""")
cursor.execute("""CREATE TABLE IF NOT EXISTS cust(cust_id INTEGER PRIMARY KEY,cust_name TEXT,city TEXT)""")
conn.commit()
cursor.execute("""CREATE TABLE IF NOT EXISTS ord(ord_id INTEGER PRIMARY KEY,cust_id INTEGER,date DATE,amount INTEGER,pay_mode TEXT CHECK (pay_mode IN ('UPI','cash')),FOREIGN KEY(cust_id) REFERENCES cust(cust_id))
""")
conn.commit()
cust_data=[(12,"ram","chennai"),(14,"sam","coimbatore"),(16,"tanu","bangalore"),(20,"ravi","coimbatore"),(23,"sam","chennai")]
ord_data=[(1,12,'2021',15000,'UPI'),(3,14,'2022',120000,'cash'),(5,16,'2020',4000,'UPI'),(7,20,'2023',50000,'cash'),(9,23,'2022',60000,'UPI')]
cursor.executemany("INSERT INTO cust VALUES(?,?,?)",cust_data)
cursor.executemany("INSERT INTO ord VALUES(?,?,?,?,?)",ord_data)
conn.commit()
cursor.execute("SELECT cust_name  from cust INNER JOIN ord ON cust.cust_id=ord.cust_id WHERE amount>100000")
rows=cursor.fetchall()
for row in rows:
  print(row)
cursor.execute("""UPDATE ord SET amount=amount*5%100 WHERE pay_mode='UPI'""")
rows=cursor.fetchall()
for row in rows:
  print(row)
cursor.execute("""DELETE FROM ord WHERE date<'2022' AND amount<5000""")
rows=cursor.fetchall()
for row in rows:
  print(row)

cursor.execute("""
SELECT
    cust.city,
    SUM(ord.amount) AS revenue,
    COUNT(ord.ord_id) AS order_count
FROM ord
JOIN cust
    ON ord.cust_id = cust.cust_id
GROUP BY cust.city
ORDER BY cust.city
""")
rows=cursor.fetchall()
for row in rows:
  print(row)

conn.commit()
conn.close()
```

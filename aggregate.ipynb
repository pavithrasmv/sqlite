```python
import sqlite3
conn=sqlite3.connect("aggerag.db")
cursor=conn.cursor()
print("database created")
cursor.execute("CREATE TABLE IF NOT EXISTS std(stud_id INTEGER PRIMARY KEY,name TEXT,marks INTEGER)")
conn.commit()
stud_data=[(101,"ram",60),(102,"sam",78),(103,"ravi",67),(104,"tanu",89),(105,"shana",88)]
cursor.executemany("""INSERT INTO std VALUES(?,?,?)""",stud_data)
conn.commit()
print("table created successfully")
cursor.execute("SELECT COUNT(*)AS tot,SUM(marks)AS tot_marks,MIN(marks)AS min_marks,AVG(marks)as avg_marks FROM std")

rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

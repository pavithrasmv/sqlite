```python
import sqlite3
conn=sqlite3.connect("aggerag.db")
cursor=conn.cursor()
print("database created")
cursor.execute("CREATE TABLE IF NOT EXISTS std(stud_id INTEGER PRIMARY KEY,name TEXT,marks INTEGER)")
conn.commit()
stud_data=[(101,"ram",60),(102,"sam",78),(103,"ravi",67),(104,"tanu",89),(105,"shana",88)]
cursor.executemany("""INSERT INTO std VALUES(?,?,?)""",stud_data)
conn.commit()
print("table created successfully")
cursor.execute("SELECT COUNT(*)AS tot,SUM(marks)AS tot_marks,MIN(marks)AS min_marks,AVG(marks)as avg_marks FROM std")

rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

```python
import sqlite3
conn=sqlite3.connect("student.db")
cursor=conn.cursor()
print("database created")
cursor.execute("""CREATE TABLE IF NOT EXISTS students(std_id INTEGER PRIMARY KEY,name TEXT,class INTEGER,marks INTEGER)""")
conn.commit()
data=[(21,"ram",10,78),(22,"sam",10,89),(23,"ravi",10,56),(24,"kavi",10,90),(25,"pami",10,88)]
cursor.executemany("INSERT INTO students VALUES(?,?,?,?)",data)
conn.commit()
cursor.execute("SELECT COUNT(*)AS tot_student,AVG(marks)AS avg_marks,MAX(marks)AS highest_score,MIN(marks)AS lowest_score FROM students")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

```python
import sqlite3
conn=sqlite3.connect("commpany_db")
cursor=conn.cursor()
print("database created")
cursor.execute("""CREATE TABLE IF NOT EXISTS employee(emp_id INTEGER PRIMARY KEY,emp_name TEXT,dept TEXT,salary INTEGER)""")
conn.commit()
print("table created")
emp_data=[(1,"ravi","IT",50000),(2,"sneha","HR",35000),(3,"manoj","IT",45000),(4,"priya","Finance",38000),(5,"karthik","Finance",42000),(6,"anu","IT",48000),(7,"vijay","HR",30000)]
cursor.executemany("INSERT INTO employee VALUES(?,?,?,?)",emp_data)
conn.commit()
cursor.execute("SELECT dept,AVG(salary)AS avg_salary FROM employee GROUP BY dept HAVING AVG(salary)>40000")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()

```

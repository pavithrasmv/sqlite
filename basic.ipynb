```python
import sqlite3

conn = sqlite3.connect("student.db", timeout=10)

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS studentss (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

students_data = [
    (1, "pavi", 29, 78),
    (2, "moni", 28, 67),
    (3, "pipi", 29, 78)
]

cursor.executemany(
    "INSERT OR IGNORE INTO studentss (id, name, age, marks) VALUES (?, ?, ?, ?)",
    students_data
)

conn.commit()

cursor.execute("SELECT * FROM studentss")

rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

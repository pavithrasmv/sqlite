```python
import sqlite3

conn = sqlite3.connect("student.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE  stud (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

students_data = [
    (1, "pavi", 29, 78),
    (2, "moni", 28, 67),
    (3, "pipi", 29, 78)
]

cursor.executemany(
    "INSERT OR IGNORE INTO stud (id, name, age, marks) VALUES (?, ?, ?, ?)",
    students_data
)
cursor.execute("UPDATE stud SET marks=75 where id =2")
conn.commit()
cursor.execute("DELETE FROM stud where id=3")
conn.commit()
cursor.execute("SELECT * FROM stud")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

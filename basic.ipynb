```python
#create a database and create a table with col:id,name,age,marks .Insert 3 student records and update mark of one student and delete
#one record and display final output .Rename and delete the database

import sqlite3

conn = sqlite3.connect("student.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE  stu (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

students_data = [
    (1, "pavi", 29, 78),
    (2, "moni", 28, 67),
    (3, "pipi", 29, 78)
]

cursor.executemany(
    "INSERT OR IGNORE INTO stu (id, name, age, marks) VALUES (?, ?, ?, ?)",
    students_data
)
cursor.execute("UPDATE stu SET marks=75 where id =2")
conn.commit()
cursor.execute("DELETE FROM stu where id=3")
conn.commit()
cursor.execute("SELECT * FROM stu")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
import os
os.rename("student.db","stu.db")
print("database renamed successfully")
os.remove("stu.db")
print("database deleted successfully")
```

```python
#Create employee databse with table employees and insert 4 records.increase salary by 10% for IT department
#delete employee earning less than 30,000 and display final table

import sqlite3
conn=sqlite3.connect("employee.db")
cursor=conn.cursor()
cursor.execute(""" CREATE TABLE emp(emp_id INTEGER PRIMARY KEY,emp_name TEXT,department TEXT,salary INTEGER)""")
emp_details=[(101,"ram","IT",25000),(103,"manu","HR",50000),(106,"sam","IT",45000),(110,"Tanu","HR",60000)]
cursor.executemany("INSERT INTO emp VALUES(?,?,?,?)",emp_details)
conn.commit()
cursor.execute("""UPDATE emp SET salary=salary*0.10 where department="IT" """)
conn.commit()
cursor.execute("DELETE FROM emp where salary<30000")
conn.commit()
cursor.execute("SELECT *from emp")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
import os
os.rename("employee.db","emp.db")
print("database renamed successfully")
os.remove("emp.db")
print("database deleted successfully")


```

```python
#A training institute wants to build a student management database with proper data integrity
import sqlite3
conn = sqlite3.connect("stu.db")
cursor = conn.cursor()
cursor.execute("""
CREATE TABLE IF NOT EXISTS stu(
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER CHECK(age >= 18),
    email TEXT UNIQUE,
    course TEXT DEFAULT 'Data Science',
    fees_paid REAL CHECK(fees_paid >= 0)
)
""")
conn.commit()
stu_data = [
    (2, "ram", 19, "ram@gmail.com", "Data Science", 20000),
    (3, "sam", 23, "sam@gmail.com", "AI", 30000),
    (4, "sanu", 24, "sanur@gmail.com", "Data Science", 23000),
    (6, "veni", 22, "veni@gmail.com", "Data analyst", 20000)
]
cursor.executemany(
    "INSERT INTO stu VALUES (?, ?, ?, ?, ?, ?)",
    stu_data
)
#test constraints:age less than 18
conn.commit()
try:
    cursor.execute("""
        INSERT INTO stu
        VALUES (5, "kiran", 16, "kiran@gmail.com", "ML", 20000)
    """)
    conn.commit()

except Exception as e:
    print("Error:", e)
cursor.execute("SELECT * FROM stu")
rows = cursor.fetchall()
for row in rows:
    print(row)
conn.close()
```

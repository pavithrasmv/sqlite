```python
#employee data cleanup:Remove employees with null or negative salaries from database
import sqlite3
conn=sqlite3.connect("employee.db")
cursor=conn.cursor()
cursor.execute("CREATE TABLE IF NOT EXISTS emp(emp_id INTEGER PRIMARY KEY,emp_name TEXT,salary INTEGER,dept TEXT)")
conn.commit()
emp_data=[(101,"supal",60000,"HR"),(102,"ram",-20000,"sales"),(103,"manu",30000,"IT"),(104,"ram",None,"IT")]
conn.commit()
cursor.executemany("INSERT INTO emp values(?,?,?,?)",emp_data)
conn.commit
cursor.execute("DELETE FROM emp where salary IS NULL or salary< 0")
print("deleted successfully")
conn.commit()
cursor.execute("SELECT * from emp")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

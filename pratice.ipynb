```python
#employee data cleanup:Remove employees with null or negative salaries from database
import sqlite3
conn=sqlite3.connect("employee.db")
cursor=conn.cursor()
cursor.execute("CREATE TABLE IF NOT EXISTS emp(emp_id INTEGER PRIMARY KEY,emp_name TEXT,salary INTEGER,dept TEXT)")
conn.commit()
emp_data=[(101,"supal",60000,"HR"),(102,"ram",-20000,"sales"),(103,"manu",30000,"IT"),(104,"ram",None,"IT")]
conn.commit()
cursor.executemany("INSERT INTO emp values(?,?,?,?)",emp_data)
conn.commit
cursor.execute("DELETE FROM emp where salary IS NULL or salary< 0")
print("deleted successfully")
conn.commit()
cursor.execute("SELECT * from emp")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

```python
#FIND DUPLICATE CUSTOMERS:Identify customers who have made multiple purchases
import sqlite3
conn=sqlite3.connect("sales.db")
cursor=conn.cursor()
cursor.execute("CREATE TABLE IF NOT EXISTS custo(cust_id INTEGER PRIMARY KEY,cust_name TEXT,produ_name TEXT)")
conn.commit()
cust_data=[(202,"pavi","laptop"),(203,"pavi","mobile"),(205,"supal","laptop"),(206,"sri","laptop"),(208,"moni","mobile")]
conn.commit()
cursor.executemany("INSERT INTO custo values(?,?,?)",cust_data)
conn.commit()
cursor.execute("SELECT cust_name ,count(*)as tot_order from custo group by cust_name HAVING count(*)>1")
conn.commit()
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

```python
#CONDITIONAL SALARY UPDATE:Update salaries :IT department+10%,HR +5%,other no change
import sqlite3
conn=sqlite3.connect("employee.db")
cursor=conn.cursor()
cursor.execute("CREATE TABLE IF NOT EXISTS emp(emp_id INTEGER PRIMARY KEY,emp_name TEXT,dept TEXT,salary INTEGER)")
conn.commit()
emp_data=[(23,"suresh","HR",60000),(28,"pavi","IT",40000),(34,"manu","IT",45000),(45,"ram","HR",50000),(46,"sam","IT",30000)]
conn.commit()
cursor.executemany("INSERT INTO emp values(?,?,?,?)",emp_data)
conn.commit()
cursor.execute("""UPDATE emp SET salary=CASE WHEN dept="HR" THEN salary *1.05 WHEN dept="IT" THEN salary *1.10 ELSE salary END""")
print("updated successfully")
conn.commit()
cursor.execute("SELECT * from emp")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```

```python
#ADVANCED FILTERING:Find employees earning more than ALL HR employees
import sqlite3

conn = sqlite3.connect("staff.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS emp_det(
    emp_name TEXT,
    salary INTEGER,
    dept TEXT
)
""")

emp_data = [
    ("suresh", 40000, "HR"),
    ("ram", 30000, "IT"),
    ("sam", 40000, "HR"),
    ("manu", 35000, "IT")
]

cursor.executemany(
    "INSERT INTO emp_det VALUES(?,?,?)",
    emp_data
)

conn.commit()

cursor.execute("""
SELECT emp_name, salary
FROM emp_det
WHERE salary > (
    SELECT MAX(salary)
    FROM emp_det
    WHERE dept = 'HR'
)
""")
cursor.execute("SELECT * from emp_det")

rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

```python
#TOP PAID EMPLOYEES BY DEPTARTMENT
import sqlite3
conn=sqlite3.connect("empdet.db")
cursor=conn.cursor()
cursor.execute("""CREATE TABLE IF NOT EXISTS staff(emp_name TEXT,dept TEXT,salary INTEGER)""")
conn.commit()
staff_det=[("pavi","IT",70000),("meena","IT",70000),("suresh","IT",50000),("Anita","HR",60000),("karthik","HR",55000)]
cursor.executemany("INSERT INTO staff values(?,?,?)",staff_det)
conn.commit()
query=""" SELECT emp_name,dept,salary RANK from(SELECT emp_name,dept,salary, DENSE_RANK() over (PARTITION BY dept ORDER BY salary DESC)as rank from staff)
where rank<=2"""
for row in cursor.execute(query):
  print(row)
conn.close()
```
